```python
print("Rows with NaN values:")
print(df[df.isnull().any(axis=1)])
# Checking for negative values in 'Experience'
df = df[df['Experience'] >= 0]
```

```python
#deleting the ID and zip code cloumn as it is not needed in visualization
if 'ID' in df.columns:
    df.drop(columns=['ID'], inplace=True)
if 'ZIP Code' in df.columns:
    df.drop(columns=['ZIP Code'], inplace=True)
df.head(10)
```

```python
print(df.describe())
pd.isna(df).any()
```

```python
# Identify non-numeric columns
non_numeric_cols = df.select_dtypes(include=['object']).columns
print("\nNon-numeric columns:", non_numeric_cols)
```

```python
for col in non_numeric_cols:
    print(f"\nChecking non-numeric values in {col}:")
    print(df[col].str.isnumeric().value_counts())
    print(df[col].loc[df[col].str.isnumeric() == False])
```

```python
# Replacing '?' with NaN
df.replace('?', pd.NA, inplace=True)
```

```python
# Convert relevant columns to numeric types
for col in non_numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
# Check for missing values
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Fill missing values with column means
df.fillna(df.mean(), inplace=True)
```

```python
# Import required library
import pandas as pd

# Load the dataset
df = pd.read_csv("credit card.csv")

# Compute Mean
mean_values = df.mean(numeric_only=True)
print("Mean values:\n", mean_values)
```

```python
import pandas as pd

# Ensure only numeric columns are used
df_numeric = df.select_dtypes(include=['number'])

# Median
median_values = df_numeric.median()

# Quantiles (Q1, Q2, Q3)
quantiles = df_numeric.quantile([0.25, 0.50, 0.75])

# Deciles (Dividing into 10 parts)
deciles = df_numeric.quantile([i/10 for i in range(1, 10)])

# Percentiles (Every 10%)
percentiles = df_numeric.quantile([i/100 for i in range(1, 101, 10)])

# Print results
print("Median:\n", median_values)
print("Quantiles:\n", quantiles)
print("Deciles:\n", deciles)
print("Percentiles:\n", percentiles)
```

```python
range_values = df.max(numeric_only=True) - df.min(numeric_only=True)
print("Range:\n", range_values)
```

```python
iqr = df.quantile(0.75, numeric_only=True) - df.quantile(0.25, numeric_only=True)
print("Interquartile Range:\n", iqr)
```

```python
interdecile_range = df.quantile(0.90, numeric_only=True) - df.quantile(0.10, numeric_only=True)
print("Interdecile Range:\n", interdecile_range)
```

```python
skewness = df.skew(numeric_only=True)
kurtosis = df.kurt(numeric_only=True)

print("Skewness:\n", skewness)
print("Kurtosis:\n", kurtosis)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Frequency Distribution
num_col = 'Income'  # Change column name accordingly
freq_distribution = df[num_col].value_counts().sort_index()
print("Frequency Distribution:\n", freq_distribution)

# Histogram
plt.figure(figsize=(8,6))
sns.histplot(df[num_col], bins=30, kde=True)
plt.title(f'Histogram of {num_col}')
plt.show()

# Relative Frequency Distribution
relative_freq = freq_distribution / freq_distribution.sum()
print("Relative Frequency Distribution:\n", relative_freq)

# Cumulative Frequency Distribution
cumulative_freq = freq_distribution.cumsum()
print("Cumulative Frequency Distribution:\n", cumulative_freq)

# Plot Cumulative Distribution
plt.figure(figsize=(8,6))
plt.plot(cumulative_freq, marker='o', linestyle='-')
plt.title(f'Cumulative Frequency Distribution of {num_col}')
plt.xlabel(num_col)
plt.ylabel('Cumulative Frequency')
plt.show()
```

```python
cat_col = 'Education'  # Change accordingly
plt.figure(figsize=(6,6))
df[cat_col].value_counts().plot.pie(autopct='%1.1f%%', cmap='viridis')
plt.title(f'Pie Chart of {cat_col}')
plt.show()
```

```python
cat1 = 'Education'  # Change accordingly
cat2 = 'Personal Loan'

cross_tab = pd.crosstab(df[cat1], df[cat2], normalize='index')
cross_tab.plot(kind='bar', stacked=True, figsize=(8,6), colormap='coolwarm')
plt.title(f'Stacked Bar Plot: {cat1} vs {cat2}')
plt.ylabel('Proportion')
plt.show()
```
