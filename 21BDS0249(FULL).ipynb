```python
cat3 = 'CreditCard'  # Choose third categorical variable

# Create a three-way contingency table
three_way_table = pd.crosstab(index=[df[cat1], df[cat2]], columns=df[cat3])
print("Three-Way Contingency Table:\n", three_way_table)
```

```python
from scipy.stats import chi2_contingency

# Row Profile (Normalization by row sum)
row_profile = three_way_table.div(three_way_table.sum(axis=1), axis=0)

# Column Profile (Normalization by column sum)
col_profile = three_way_table.div(three_way_table.sum(axis=0), axis=1)

# Relative Frequency
relative_freq = three_way_table / three_way_table.sum().sum()

# Chi-Square Test
chi2, p, dof, expected = chi2_contingency(three_way_table)

print("Row Profile:\n", row_profile)
print("Column Profile:\n", col_profile)
print("Relative Frequency Table:\n", relative_freq)
print(f"Chi-Square Statistic: {chi2}, p-value: {p}")
```

```python
# Select numerical features
data = df[['Income', 'CCAvg']].values
```

```python
from sklearn.preprocessing import StandardScaler

# Scale the data
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Compute WCSS (Within-Cluster Sum of Squares)
wcss = []
for i in range(1, 11):  # Checking clusters from 1 to 10
    kmeans = KMeans(n_clusters=i, random_state=123, n_init=10)
    kmeans.fit(data_scaled)
    wcss.append(kmeans.inertia_)

# Plot the Elbow Graph
plt.figure(figsize=(8,6))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.title('Elbow Method for Optimal Clusters')
plt.show()
```

```python
# Apply K-Means with 5 clusters (change this based on elbow method result)
kmeans = KMeans(n_clusters=5, random_state=123, n_init=10)
kmeans_result = kmeans.fit_predict(data_scaled)

# Add cluster labels to dataset
df['KMeans_Cluster'] = kmeans_result

# Plot the K-Means Clustering Results
plt.figure(figsize=(8,6))
plt.scatter(data_scaled[:, 0], data_scaled[:, 1], c=kmeans_result, cmap='viridis', alpha=0.6)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=200, c='red', marker='X', label='Centroids')
plt.xlabel("Income (Scaled)")
plt.ylabel("CCAvg (Scaled)")
plt.title("K-Means Clustering")
plt.legend()
plt.show()
```

```python
from sklearn.cluster import DBSCAN

# Apply DBSCAN with epsilon=0.5 and min_samples=5
dbscan = DBSCAN(eps=0.5, min_samples=5)
dbscan_labels = dbscan.fit_predict(data_scaled)

# Add DBSCAN labels to dataset
df['DBSCAN_Cluster'] = dbscan_labels

# Visualize DBSCAN results
plt.figure(figsize=(8,6))
plt.scatter(data_scaled[:, 0], data_scaled[:, 1], c=dbscan_labels, cmap='coolwarm', alpha=0.6)
plt.xlabel("Income (Scaled)")
plt.ylabel("CCAvg (Scaled)")
plt.title("DBSCAN Clustering")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
numeric_cols = df.select_dtypes(include=['number'])
```

```python
# Standardize the data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(numeric_cols)

# Apply PCA
pca = PCA()
pca_results = pca.fit_transform(scaled_data)

# Get eigenvalues
eigenvalues = pca.explained_variance_ratio_
print("Eigenvalues:\n", eigenvalues)

# Get component loadings
loadings = pca.components_
print("Component Loadings:\n", loadings)
```

```python
plt.figure(figsize=(8,6))
plt.plot(range(1, len(eigenvalues) + 1), eigenvalues, marker='o', linestyle='--')
plt.xlabel('Principal Component')
plt.ylabel('Variance Explained')
plt.title('Scree Plot for PCA')
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(pca_results[:, 0], pca_results[:, 1], alpha=0.5)
plt.xlabel("First Principal Component")
plt.ylabel("Second Principal Component")
plt.title("PCA Biplot")
plt.show()
```

```python
# Select the top components based on Scree Plot
n_comp = 3

pca_selected = PCA(n_components=n_comp)
pca_data = pca_selected.fit_transform(scaled_data)

# Visualize selected PCA components
plt.figure(figsize=(8,6))
plt.scatter(pca_data[:, 0], pca_data[:, 1], alpha=0.5)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Selected PCA Components")
plt.show()

# Print Summary
print(pd.DataFrame(pca_selected.explained_variance_ratio_, columns=['Variance Explained']))
```

```python
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Select features for Simple Linear Regression
X = df[['Income']]  # Independent variable
y = df['CCAvg']  # Dependent variable

# Split dataset into Training (2/3) and Testing (1/3)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=1/3, random_state=123)

# Fit Simple Linear Regression Model
reg1 = LinearRegression()
reg1.fit(X_train, y_train)

# Print model summary
print("Intercept:", reg1.intercept_)
print("Coefficient:", reg1.coef_)

# Predict Test Set Results
y_pred = reg1.predict(X_test)

# Visualize results
plt.figure(figsize=(8,6))
plt.scatter(X_test, y_test, color='blue', label="Actual Data")
plt.plot(X_test, y_pred, color='red', linewidth=2, label="Regression Line")
plt.xlabel("Income")
plt.ylabel("Credit Card Average Spending")
plt.title("Simple Linear Regression")
plt.legend()
plt.show()

# Check residual errors
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print("Mean Squared Error:", mse)
print("R-squared:", r2)
```

```python
# Select multiple independent variables & log transform 'Income'
X_multi = df[['Education', 'Mortgage', 'Securities Account', 'CD Account', 'CreditCard']]
X_multi['log_Income'] = np.log(df['Income'])

y_multi = df['CCAvg']  # Dependent variable

# Split into Train & Test sets
X_train, X_test, y_train, y_test = train_test_split(X_multi, y_multi, test_size=1/3, random_state=123)

# Fit Multiple Linear Regression
reg2 = LinearRegression()
reg2.fit(X_train, y_train)

# Print Model Summary
print("Intercept:", reg2.intercept_)
print("Coefficients:", reg2.coef_)

# Predict Test Set Results
y_pred_multi = reg2.predict(X_test)

# Compare Predicted vs Residuals
residuals = y_test - y_pred_multi

# Create DataFrame for Predicted & Residuals
result_df = pd.DataFrame({'Predicted': y_pred_multi, 'Residuals': residuals})

print(result_df.head())

# Visualize Regression Results using QQ Plot
import scipy.stats as stats

plt.figure(figsize=(8,6))
stats.probplot(residuals, dist="norm", plot=plt)
plt.title("QQ Plot for Residuals")
plt.show()
```
